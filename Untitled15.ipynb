```python
from transformers import pipeline
import gradio as gr
```

```python
qa_pipeline = pipeline("question-answering")
```

```python
def answer_question(context, question):

    result = qa_pipeline(
        question=question,
        context=context
    )

    return result['answer']
```

```python
interface = gr.Interface(
    fn=answer_question,
    inputs=[
    gr.Textbox(lines=8, label="Enter Paragraph"),
    gr.Textbox(label="Enter Question")
],
     outputs="text",
    title="GenAI QA App",
    description="Ask anything!"
)
```

```python
interface.launch()
```
